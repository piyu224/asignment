```python
# 1.

import pandas as pd

january = pd.DataFrame({
    "OrderID": [101, 102, 103],
    "Restaurant": ["Dominos", "KFC", "Pizza Hut"],
    "Amount": [500, 350, 600]
})

february = pd.DataFrame({
    "OrderID": [104, 105, 106],
    "Restaurant": ["Burger King", "McDonalds", "Subway"],
    "Amount": [450, 300, 400]
})

all_orders = pd.concat([january, february], ignore_index=True)

print(all_orders)
```

```python
# 2.

import pandas as pd

products = pd.DataFrame({
    "ProductID": [101, 102, 103, 104],
    "ProductName": ["Laptop", "Mobile", "Headphones", "Keyboard"],
    "Price": [50000, 20000, 1500, 1200]
})

ratings = pd.DataFrame({
    "Rating": [5, 4, 5, 3]
})

result = pd.concat([products, ratings], axis=1)
print(result)
```

```python
# 3.

import pandas as pd

df = pd.DataFrame({
    "User": [
        "Rahul", "Rahul", "Rahul", "Rahul", "Rahul", "Rahul", "Rahul",
        "Priya", "Priya", "Priya", "Priya", "Priya", "Priya", "Priya",
        "Amit", "Amit", "Amit", "Amit", "Amit", "Amit", "Amit"
    ],
    "Day": [
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"
    ],
    "Streams": [
        120, 150, 100, 180, 200, 250, 220,
        100, 130, 140, 160, 180, 210, 190,
        80, 110, 90, 120, 150, 170, 160
    ]
})

result = df.pivot(
    index="User",
    columns="Day",
    values="Streams"
)

print(result)
```

```python
# 4.

import pandas as pd

df = pd.DataFrame({
    "restaurant": [
        "Dominos", "Dominos", "Dominos",
        "KFC", "KFC", "KFC",
        "Pizza Hut", "Pizza Hut", "Pizza Hut"
    ],
    "month": [
        "January", "January", "February",
        "January", "February", "February",
        "January", "January", "February"
    ],
    "total_amount": [
        500, 700, 600,
        400, 500, 600,
        800, 600, 700
    ]
})

result = df.pivot_table(
    index="restaurant",
    columns="month",
    values="total_amount",
    aggfunc="mean"
)
print(result)
```

```python
# 5.

import pandas as pd

jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")

all_sales = pd.concat([jan, feb, mar], ignore_index=True)

total_sales = all_sales.groupby("product")["sales"].sum()

print(total_sales)
```
