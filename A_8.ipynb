```python
# 1. 

import numpy as np

teams = np.array([
    "CSK", "MI", "RCB", "CSK", "KKR",
    "MI", "GT", "RCB", "RR", "KKR"
])
unique_teams = np.unique(teams)

print("Unique IPL Teams:")
print(unique_teams)
```

```python
# 2. 

import numpy as np

ratings = np.array([4.5, 4.0, np.nan, 3.8, np.nan, 4.7, 5.0])

missing = np.isnan(ratings)

missing_count = np.sum(missing)

print("Ratings:", ratings)
print("Missing ratings:", missing_count)
```

```python
# 3.

import numpy as np

prices = np.array([50, 150, 500, 1200, 800, 1500, 90])
clipped_prices = np.clip(prices, 100, 1000)

print("Original Prices:", prices)
print("Clipped Prices:", clipped_prices)
```

```python
# 4.

import numpy as np

views = np.array([1000, 2500, np.nan, 5000, np.inf, 7500, np.nan, np.inf])

max_finite = np.max(views[np.isfinite(views)])
views[np.isnan(views)] = 0

views[np.isinf(views)] = max_finite

print("Updated Views:")
print(views)
```

```python
# 5.

import numpy as np
numbers = np.array([11, 24, 35, 42, 53, 60, 71, 88])

even_indices = np.where(numbers % 2 == 0)

print("Array:", numbers)
print("Indices of even numbers:", even_indices[0])
```
