```python
# 1.

import pandas as pd

df = pd.DataFrame({"username": [ "CoolDude","foodieQueen","TravelStar","Tech_Guru","FashionLover" ]
})

df["username"] = df["username"].str.lower()
print(df)
```

```python
# 2.

import pandas as pd

songs = pd.Series(["Shape of You Remix","Blinding Lights","Perfect Remix","Kesariya Remix","Tum Hi Ho"])

cleaned_songs = songs.str.replace("Remix", "", regex=False)

print(cleaned_songs)
```

```python
# 3.

import pandas as pd

df = pd.DataFrame({
    "review": [
        "The food was delicious and delivery was fast",
        "Good taste and great service",
        "Delivery was very late",
        "Amazing food and friendly staff",
        "Fast Delivery and tasty food"
    ]
})
result = df[df["review"].str.contains("delivery", case=False, na=False)]
print(result)
```

```python
# 4.

import pandas as pd

df = pd.DataFrame({
    "Category": [
        "Electronics/Mobiles/Smartphones",
        "Home/Kitchen/Appliances",
        "Fashion/Men/Shirts",
        "Books/Fiction/Novels",
        "Beauty/Makeup/Lipstick"
    ]
})

df["MainCategory"] = df["Category"].str.split("/").str[0]
print(df)
```

```python
# 5.

import pandas as pd

df = pd.DataFrame({
    "Title": [
        "Python Tutorial",
        "Data Science Course",
        "Learn Pandas",
        "YouTube Tips"
    ]
})

df["Title"] = df["Title"].str.upper().str.replace(" ", "_", regex=False)
print(df)
```
