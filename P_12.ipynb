```python
# 1.

import pandas as pd

df = pd.read_csv("ipl_matches.csv")

print("IPL Dataset:")
print(df.head())

teams = pd.concat([df["Team1"], df["Team2"]])

result = teams.value_counts().reset_index()

result.columns = ["Team", "Total Matches"]

print("\nTeam-wise Total Matches:")
print(result)
```

```python
# 2.

import pandas as pd

data = {
    "city": ["Mumbai", "Mumbai", "Mumbai", "Delhi", "Delhi", "Delhi"],
    "restaurant": ["Dominos", "Dominos", "McDonalds", "Dominos", "McDonalds", "McDonalds"],
    "order_amount": [500, 700, 400, 600, 500, 700]
}
df = pd.DataFrame(data)

print("Zomato Food Orders:")
print(df)

result = df.groupby(["city", "restaurant"]).agg(
    average_order_amount=("order_amount", "mean")
).reset_index()

print("\nAverage Order Amount  Restaurant in City:")
print(result)
```

```python
# 3.

import pandas as pd

# Create Flipkart-style product sales DataFrame
data = {
    "category": ["Electronic", "Electronic", "Clothing", "Clothing",
                 "Electronics", "Clothing"],
    "region": ["North", "South", "North", "South", "North", "North"],
    "sales": [50000, 70000, 30000, 40000, 60000, 35000],
    "units_sold": [50, 70, 30, 40, 60, 35]
}
df = pd.DataFrame(data)

print("Flipkart Product Sales:")
print(df)

result = df.groupby(["category", "region"]).agg(
    total_sales=("sales", "sum"),
    average_units_sold=("units_sold", "mean")
).reset_index()

print("\nCategory and Region-wise Sales:")
print(result)
```

```python
# 4.

import pandas as pd

data = {
    "region": ["North", "North", "South", "South", "West", "West"],
    "revenue": [50000, 70000, 40000, 60000, 30000, 50000]
}
df = pd.DataFrame(data)

result = df.groupby("region").agg({
    "revenue": ["sum", "max", "min"]
})

print("Revenue Summary by Region:")
print(result)
```

```python
# 5.

import pandas as pd

data = {
    "artist": ["Taylor Swift", "Drake", "Ed Sheeran", "Taylor Swift",
               "Drake", "The Weeknd"],
    "genre": ["Pop", "Hip-Hop", "Pop", "Pop", "Hip-Hop", "R&B"],
    "streams": [5000000, 7000000, 4000000, 6000000, 8000000, 3000000]
}

df = pd.DataFrame(data)

result = df.groupby("genre").agg({
    "streams": ["sum", "mean"]
})
result.columns = ["total_streams", "average_streams"]

result = result.sort_values("total_streams", ascending=False)

print("Genre-wise Spotify Streams:")
print(result)
```
