```python
# 1. Create a Python class called Product with a private attribute _price. Initialize _price in the constructor and write a method to display its value.

class Product:
    def __init__(self, price):
        self._price = price

    def display_price(self):
        print("Price:", self._price)

product1 = Product(500)

product1.display_price()
```

```python
# 2. Add getter and setter methods for the _price attribute in your Product class to safely access and update the price. Make sure the setter prevents setting a negative price.
#   <br><br><em><strong>Hint:</strong> Raise a ValueError if the new price is less than zero.</em>

class Product:
    def __init__(self, price):
        self._price = price

    # Getter
    def get_price(self):
        return self._price

    # Setter
    def set_price(self, price):
        if price < 0:
            raise ValueError("Price cannot be negative")
        self._price = price


product1 = Product(500)

print("Old Price:", product1.get_price())

product1.set_price(800)

print("New Price:", product1.get_price())
```

```python
# 3. Build a class called Playlist that has a private attribute _songs (a list of song names). 
#    Write methods to add a song, remove a song, and get the current list of songs using proper encapsulation.

class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)

    def get_songs(self):
        return self._songs


playlist = Playlist()

playlist.add_song("Kesariya")
playlist.add_song("Apna Bana Le")
playlist.add_song("Tum Hi Ho")

print("Songs:", playlist.get_songs())

playlist.remove_song("Apna Bana Le")

print("After Removing:", playlist.get_songs())
```

```python
# 4. Create an abstract class PaymentMethod with an abstract method pay(amount). Then, create two subclasses: UPI and CreditCard, each implementing the pay method with a print statement 
#    showing how the payment would be processed.<br><br><em><strong>Hint:</strong> Use the abc module for abstraction.</em>

from abc import ABC, abstractmethod

class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class UPI(PaymentMethod):

    def pay(self, amount):
        print(f"Payment of ₹{amount} processed through UPI")

class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Payment of ₹{amount} processed through Credit Card")

upi_payment = UPI()
card_payment = CreditCard()

upi_payment.pay(500)
card_payment.pay(1000)
```
