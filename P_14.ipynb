```python
# 3.

import pandas as pd
import numpy as np

data = {
    "Match": ["Match 1", "Match 2", "Match 3", "Match 4", "Match 5"],
    "Team1": ["CSK", "RCB", "MI", "CSK", "KKR"],
    "Team2": ["MI", "KKR", "GT", "RCB", "MI"],
    "Winner": ["CSK", "KKR", "GT", "RCB", "MI"],
    "win_by_runs": [20, 0, 0, 0, 15],
    "win_by_wickets": [0, 6, 7, 5, 0]
}

ipl_df = pd.DataFrame(data)

conditions = [
    ipl_df["win_by_runs"] > 0,
    ipl_df["win_by_wickets"] > 0
]

choices = ["Runs", "Wickets"]

ipl_df["match_margin_type"] = np.select(
    conditions,
    choices,
    default="Tie/No Result"
)
display(ipl_df)
```

```python
# 4.

import pandas as pd

team_data = {
    "team": ["CSK", "MI", "RCB", "KKR", "GT"],
    "home_city": ["Chennai", "Mumbai", "Bangalore", "Kolkata", "Ahmedabad"]
}

team_df = pd.DataFrame(team_data)

ipl_df = ipl_df.merge(
    team_df,
    left_on="Team1",
    right_on="team",
    how="left"
)

ipl_df = ipl_df.rename(columns={"home_city": "Team1_Home_City"})

ipl_df = ipl_df.drop(columns=["team"])

ipl_df = ipl_df.merge(
    team_df,
    left_on="Team2",
    right_on="team",
    how="left"
)

ipl_df = ipl_df.rename(columns={"home_city": "Team2_Home_City"})

ipl_df = ipl_df.drop(columns=["team"])

display(ipl_df)
```

```python
# 5.

import pandas as pd

ipl_df["Season"] = [2023, 2023, 2023, 2023, 2023]

pivot = pd.pivot_table(
    ipl_df,
    index="Winner",
    columns="Season",
    values="Match",
    aggfunc="count",
    fill_value=0
)

print("Total Matches Won by Each Team per Season:")
display(pivot)

top_3 = (
    pivot.stack()
    .reset_index(name="Wins")
    .sort_values("Wins", ascending=False)
    .head(3)
)

print("Top 3 Teams with Most Wins in a Single Season:")
display(top_3)
```
