```python
# 1. 

import pandas as pd

df = pd.read_csv("ipl_players.csv")

print(df.columns)

missing_age = df[df['age'].isna()]

print("Players with missing age:")
print(missing_age)
```

```python
# 2. 

import pandas as pd

orders = pd.DataFrame({
    'order_id': ['ORD001', 'ORD002', 'ORD003', 'ORD004', 'ORD005'],
    'customer_name': ['Rahul', 'Priya', 'Amit', 'Neha', 'Karan'],
    'delivery_rating': [5, None, 4, None, 3]
})

print("Original DataFrame:")
print(orders)

cleaned_orders = orders.dropna(subset=['delivery_rating'])

print("\nCleaned DataFrame:")
print(cleaned_orders)
```

```python
# 3. 

import pandas as pd

reviews = pd.DataFrame({
    'product_name': [
        'Laptop',
        'Smartphone',
        'Headphones',
        'Smart Watch',
        'Tablet',
        'Keyboard'
    ],
    'rating': [4.5, 4.0, None, 5.0, None, 3.5]
})

print("Original DataFrame:")
print(reviews)

mean_rating = reviews['rating'].mean()

reviews['rating'] = reviews['rating'].fillna(mean_rating)

print("\nUpdated DataFrame:")
print(reviews)
```

```python
# 4.

import pandas as pd

df = pd.read_csv("spotify_playlist.csv")

print("Original DataFrame:")
print(df)

median_duration = df['duration'].median()

df['duration'] = df['duration'].fillna(median_duration)

print("\nCleaned DataFrame:")
print(df)

df.to_csv("spotify_playlist_cleaned.csv", index=False)

print("\nCleaned file saved successfully!")
```

```python
# 5.

import pandas as pd

instagram = pd.DataFrame({
    'username': ['user1', 'user2', 'user3', 'user4', 'user5'],
    'followers': [1200, None, 5400, None, 3200]
})

print("Original DataFrame:")
print(instagram)

instagram['followers'] = instagram['followers'].fillna(0)

print("\nUpdated DataFrame:")
print(instagram)
```
