```python
# 1. 

import numpy as np

prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print("Prices:", prices)
print("Data Type:", prices.dtype)
```

```python
# 2. 

positive_prices = prices[prices > 0]

average_price = positive_prices.mean()

prices[prices <= 0] = average_price

print("Cleaned Prices:", prices)
```

```python
# 3. 

import numpy as np

prices = np.array([299, 499, 799, 819, 1599, 819, 899])

# Quantity of each item
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

# Calculate bill for each item
total_bill = prices * quantities

print("Cleaned Prices:", prices)
print("Quantities:", quantities)
print("Total Bill for Each Item:", total_bill)
```

```python
# 4. 

import numpy as np
prices = np.array([299, 499, 799, 819, 1599, 819, 899])

quantities = np.array([2, 1, 3, 4, 2, 1, 5])

item_bill = prices * quantities

print("----- Summary Report -----")
print("Minimum Price:", prices.min())
print("Maximum Price:", prices.max())
print("Average Price:", prices.mean())
print("Total of Cleaned Prices:", prices.sum())
print("Total Bill for All Items:", item_bill.sum())
```

```python
# 5. 

import numpy as np

prices = np.array([299, 499, 799, 819, 1599, 819, 899])

unique_prices = np.unique(prices)

number_of_unique_prices = len(unique_prices)

print("Unique Prices:", unique_prices)
print("Number of Unique Prices:", number_of_unique_prices)
```
