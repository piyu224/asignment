```python
# 1. Write a recursive function in Python called reverse_string(s) that takes a string and returns it reversed (e.g., 'hello' becomes 'olleh').

def reverse_string(s):
    if s == "":
        return s
    return reverse_string(s[1:]) + s[0]

print(reverse_string("piyu"))
```

```python
# 2. Build a recursive function sum_playlist_durations(durations) that takes a list of song durations (in seconds)
# and returns the total duration, similar to how Spotify totals a playlist.

def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    
    return durations[0] + sum_playlist_durations(durations[1:])

durations = [180, 200, 240, 300]
total = sum_playlist_durations(durations)
print("Total duration:", total, "seconds")
```

```python
# 4. Create a recursive function count_likes(posts) that takes a nested dictionary representing Instagram posts and their replies (each with a 'likes' key), and returns the total number of likes across all posts and replies.<br><br><em><strong>Hint:</strong> 
#    Each reply can itself have more replies, so use recursion to sum likes at all levels.</em>

posts = [
    {
        "likes": 100,
        "replies": [
            {
                "likes": 20,
                "replies": []
            },
            {
                "likes": 30,
                "replies": [
                    {
                        "likes": 10,
                        "replies": []
                    }
                ]
            }
        ]
    },
    {
        "likes": 50,
        "replies": []
    }
]
```

```python
def count_likes(posts):
    total = 0

    for post in posts:
        total += post["likes"]

        if "replies" in post:
            total += count_likes(post["replies"])

    return total


print("Total likes:", count_likes(posts))
```
