```python
# 1. 

import numpy as np

zomato = np.array([4.5, 4.2, 3.8, 4.7, 4.0])
swiggy = np.array([4.1, 3.9, 4.6, 4.3, 4.8])

combined = np.concatenate((zomato, swiggy))

print("Zomato Ratings:", zomato)
print("Swiggy Ratings:", swiggy)
print("Combined Ratings:", combined)
```

```python
# 4. 

import numpy as np
chat = np.array([
    [20, 15, 25, 18, 30, 22, 10],   # User 1
    [10, 12, 15, 20, 18, 15, 10],   # User 2
    [25, 20, 30, 22, 35, 28, 20]    # User 3
])

print("Original Chat Data:")
print(chat)

new_user = np.array([5, 8, 10, 7, 6, 9, 5])
chat = np.insert(chat, 3, new_user, axis=0)

print("\nAfter Adding New User:")
print(chat)
totals = np.sum(chat, axis=1)

print("\nTotal Messages by Each User:", totals)
least_user = np.argmin(totals)
chat = np.delete(chat, least_user, axis=0)

print("\nAfter Removing User with Least Messages:")
print(chat)
```
