```python
# 1. 

import pandas as pd

df = pd.DataFrame({
    "ProductName": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Price": [50000, 20000, 1500, 1200, 700],
    "Qty": [2, 3, 5, 4, 6]
})

df.to_csv("flipkart_sales.csv", index=False)
```

```python
df = pd.read_csv("flipkart_sales.csv")

df["TotalValue"] = df["Price"] * df["Qty"]

print(df)
```

```python
# 2. 

import pandas as pd

df = pd.DataFrame({
    "Item": ["Pizza", "Burger", "Pasta", "Sandwich"],
    "Price": [200, 150, 180, 100],
    "Qty": [2, 3, 1, 4]
})

df["DiscountedPrice"] = df["Price"].apply(lambda x: x * 0.90)

print(df)
```

```python
# 3.

import pandas as pd

df = pd.DataFrame({
    "PostID": [101, 102, 103, 104],
    "Likes": [1500, 2500000, 850, 120000],
    "Comments": [120, 15000, 750, 250000]
})

def format_number(x):
    if x >= 1000000:
        return str(round(x / 1000000, 1)) + "M"
    elif x >= 1000:
        return str(round(x / 1000, 1)) + "K"
    else:
        return str(x)

df[["Likes", "Comments"]] = df[["Likes", "Comments"]].map(format_number)

print(df)
```

```python
# 4.

import pandas as pd

df = pd.DataFrame({
    "Restaurant": ["Dominos", "KFC", "McDonalds", "Burger King"],
    "Price": [200, 150, 180, 120],
    "Qty": [2, 3, 1, 4],
    "DeliveryCharge": [30, 40, 20, 25]
})

df["FinalAmount"] = df.apply(
    lambda x: (x["Price"] * x["Qty"]) + x["DeliveryCharge"],
    axis=1
)
print(df)
```

```python
# 5.

import pandas as pd

df = pd.DataFrame({
    "OrderID": [101, 102, 103, 104],
    "Product": ["Kurti", "Jeans", "Shoes", "T-Shirt"],
    "Price": [800, 1200, 2000, 600],
    "Qty": [2, 1, 1, 3]
})

df["GST"] = df["Price"] * df["Qty"] * 0.05

print(df)
```
