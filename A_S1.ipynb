```python
# Task 2: Restaurant Menu Manager

# Step 1: Create menu dictionary

menu = {
    "Paneer Burger": {
        "price": 180,
        "category": "Snacks"
    },
    "Masala Dosa": {
        "price": 90,
        "category": "Breakfast"
    },
    "Veg Pizza": {
        "price": 250,
        "category": "Main Course"
    },
    "French Fries": {
        "price": 120,
        "category": "Snacks"
    },
    "Paneer Tikka": {
        "price": 220,
        "category": "Starter"
    },
    "Gulab Jamun": {
        "price": 80,
        "category": "Dessert"
    }
}

# Step 2: Function to view all menu items

def view_all_items():

    print("\n----------- MENU -----------")
    print("No.  Dish Name          Price    Category")
    print("------------------------------------------")

    number = 1

    for dish, details in menu.items():

        print(
            number,
            dish,
            "      Rs", details["price"],
            "     ", details["category"]
        )

        number += 1

# Step 3: Function to filter items by category

def filter_by_category():

    category = input("Enter category: ")

    print("\nItems in", category, "category:")
    
    found = False

    for dish, details in menu.items():

        if details["category"].lower() == category.lower():

            print(
                dish,
                "- Rs",
                details["price"]
            )

            found = True

    if found == False:
        print("No items found in this category.")

# Step 4: Function to search dish by name

def search_dish():

    dish_name = input("Enter dish name: ")

    if dish_name in menu:

        print("\nDish Found!")
        print("Dish:", dish_name)
        print("Price: Rs", menu[dish_name]["price"])

    else:

        print("Dish not found.")

# Step 5: Main loop

while True:

    print("\n===== RESTAURANT MENU MANAGER =====")
    print("1. View All Items")
    print("2. Filter Items by Category")
    print("3. Search Dish by Name")
    print("0. Exit")

    choice = input("Enter your choice: ")
# Option 1

    if choice == "1":

        view_all_items()

# Option 2

    elif choice == "2":

        filter_by_category()

# Option 3

    elif choice == "3":

        search_dish()

# Exit
    elif choice == "0":

        print("Thank you! Program ended.")
        break


    # Invalid choice

    else:

        print("Invalid choice. Please try again.")
```

```python
# Task 3: Order History File Logger

import json

def load_orders():

    try:
        with open("orders.json", "r") as file:
            orders = json.load(file)

        if isinstance(orders, list):
            return orders

        elif isinstance(orders, dict):
            return [orders]

        else:
            return []

    except FileNotFoundError:
        return []

    except json.JSONDecodeError:
        return []

def save_orders(orders):

    with open("orders.json", "w") as file:
        json.dump(orders, file, indent=4)

def add_order():

    print("\n----- Add New Order -----")

    customer_name = input("Enter customer name: ")

    items_input = input("Enter items (comma-separated): ")

    items = [item.strip() for item in items_input.split(",")]

    try:
        total_amount = float(input("Enter total amount: "))

    except ValueError:
        print("Error: Total amount must be a number.")
        return

    status = input("Enter order status: ")

    orders = load_orders()

    new_order = {
        "customer_name": customer_name,
        "items": items,
        "total_amount": total_amount,
        "status": status
    }

    orders.append(new_order)

    save_orders(orders)

    print("\nOrder saved successfully!")


def view_orders():

    print("\n----- Order History -----")

    try:

        with open("orders.json", "r") as file:
            orders = json.load(file)

    except FileNotFoundError:

        print("No orders found.")
        return

    except json.JSONDecodeError:

        print("Error: orders.json contains invalid data.")
        return

    # If dictionary, convert it to list
    if isinstance(orders, dict):
        orders = [orders]

    if len(orders) == 0:

        print("No orders available.")
        return

    for i, order in enumerate(orders, start=1):

        print("\nOrder", i)
        print("-------------------------")
        print("Customer:", order["customer_name"])
        print("Items:", ", ".join(order["items"]))
        print("Total Amount: Rs", order["total_amount"])
        print("Status:", order["status"])

while True:

    print("\n==============================")
    print("   FOOD DELIVERY APP")
    print("==============================")
    print("1. Add Order")
    print("2. View Orders")
    print("0. Exit")
    print("==============================")

    choice = input("Enter your choice: ")


    if choice == "1":

        add_order()

    elif choice == "2":

        view_orders()

    elif choice == "0":

        print("\nThank you! Program ended.")
        break

    
    else:

        print("\nInvalid choice. Please enter 1, 2 or 0.")
```

```python
# Task 4 — Delivery Rider OOP System

import csv
import os

class Rider:

    def __init__(self, rider_id, name, status="Available", total_deliveries=0):

        self.rider_id = rider_id
        self.name = name
        self.status = status
        self.total_deliveries = total_deliveries

    def assign_order(self, order_id):

        self.status = "On Delivery"

        print(
            "Order", order_id,
            "assigned to rider", self.name
        )

    def complete_delivery(self):

        self.total_deliveries += 1
        self.status = "Available"

        print(
            "Delivery completed by",
            self.name
        )

    def display_info(self):

        print("\n-------------------------")
        print("Rider ID:", self.rider_id)
        print("Name:", self.name)
        print("Status:", self.status)
        print("Total Deliveries:", self.total_deliveries)
        print("-------------------------")

riders = []

if os.path.exists("riders.csv"):

    with open("riders.csv", "r", newline="") as file:

        reader = csv.DictReader(file)

        for row in reader:

            rider = Rider(
                int(row["rider_id"]),
                row["name"],
                row["status"],
                int(row["total_deliveries"])
            )

            riders.append(rider)

    print("Riders loaded from riders.csv")

if len(riders) == 0:

    rider1 = Rider(1, "Rahul")
    rider2 = Rider(2, "Amit")
    rider3 = Rider(3, "Neha")

    riders.append(rider1)
    riders.append(rider2)
    riders.append(rider3)

def find_rider(rider_id):

    for rider in riders:

        if rider.rider_id == rider_id:
            return rider

    return None

def display_all_riders():

    print("\n===== ALL RIDERS =====")

    for rider in riders:

        rider.display_info()

while True:

    print("\n==============================")
    print("     DELIVERY RIDER SYSTEM")
    print("==============================")
    print("1. Display All Riders")
    print("2. Assign Order")
    print("3. Complete Delivery")
    print("0. Exit")
    print("==============================")


    choice = input("Enter your choice: ")

    if choice == "1":

        display_all_riders()

    elif choice == "2":

        try:

            rider_id = int(input("Enter Rider ID: "))

            rider = find_rider(rider_id)

            if rider is not None:

                order_id = input("Enter Order ID: ")

                rider.assign_order(order_id)

            else:

                print("Rider not found.")


        except ValueError:

            print("Please enter a valid Rider ID.")

    elif choice == "3":

        try:

            rider_id = int(input("Enter Rider ID: "))

            rider = find_rider(rider_id)

            if rider is not None:

                rider.complete_delivery()

            else:

                print("Rider not found.")


        except ValueError:

            print("Please enter a valid Rider ID.")

    elif choice == "0":

        with open("riders.csv", "w", newline="") as file:

            fieldnames = [
                "rider_id",
                "name",
                "status",
                "total_deliveries"
            ]

            writer = csv.DictWriter(
                file,
                fieldnames=fieldnames
            )

            writer.writeheader()


            for rider in riders:

                writer.writerow({
                    "rider_id": rider.rider_id,
                    "name": rider.name,
                    "status": rider.status,
                    "total_deliveries": rider.total_deliveries
                })


        print("\nRider data saved to riders.csv")
        print("Thank you! Program ended.")

        break

    else:

        print("Invalid choice. Please try again.")
```

```python
# SEC C
# Mini Project: Food Delivery Order Management Console

import json
import os

class Order:

    def __init__(
        self,
        order_id,
        customer_name,
        items,
        total_amount,
        status="Pending"
    ):
        self.order_id = order_id
        self.customer_name = customer_name
        self.items = items
        self.total_amount = total_amount
        self.status = status

    def to_dict(self):

        return {
            "order_id": self.order_id,
            "customer_name": self.customer_name,
            "items": self.items,
            "total_amount": self.total_amount,
            "status": self.status
        }

    @classmethod
    def from_dict(cls, data):

        return cls(
            data["order_id"],
            data["customer_name"],
            data["items"],
            data["total_amount"],
            data.get("status", "Pending")
        )

orders = []

def load_orders():

    global orders

    try:

        with open("orders.json", "r") as file:

            data = json.load(file)

        # Check JSON is a list
        if not isinstance(data, list):

            print("Invalid orders.json format.")
            orders = []
            return

        orders = []

        for order_data in data:

            required_keys = [
                "order_id",
                "customer_name",
                "items",
                "total_amount"
            ]

            if all(key in order_data for key in required_keys):

                order = Order.from_dict(order_data)

                orders.append(order)

            else:

                print("Skipped an invalid order in orders.json.")

        print("Orders loaded successfully.")


    except FileNotFoundError:

        print("orders.json not found.")
        print("Starting with an empty order list.")

        orders = []


    except json.JSONDecodeError:

        print("Error: orders.json contains invalid JSON.")

        print("Starting with an empty order list.")

        orders = []

def save_orders():

    data = []

    for order in orders:

        data.append(order.to_dict())

    with open("orders.json", "w") as file:

        json.dump(
            data,
            file,
            indent=4
        )

def generate_order_id():

    if len(orders) == 0:

        return 1

    max_id = 0

    for order in orders:

        if order.order_id > max_id:

            max_id = order.order_id

    return max_id + 1

def place_order():

    print("\n================================")
    print("        PLACE NEW ORDER")
    print("================================")

    while True:

        customer_name = input(
            "Enter customer name: "
        ).strip()

        if customer_name == "":

            print("Error: Customer name cannot be empty.")

        else:

            break

    while True:

        items_input = input(
            "Enter items (comma-separated): "
        ).strip()

        if items_input == "":

            print("Error: Please enter at least one item.")

        else:

            items = []

            for item in items_input.split(","):

                item = item.strip()

                if item != "":

                    items.append(item)

            if len(items) == 0:

                print("Error: Please enter valid items.")

            else:

                break

    while True:

        try:

            total_amount = float(
                input("Enter total amount: ")
            )

            if total_amount < 0:

                print(
                    "Error: Amount cannot be negative."
                )

            else:

                break

        except ValueError:

            print(
                "Error: Please enter a numeric amount."
            )

    order_id = generate_order_id()

    new_order = Order(
        order_id,
        customer_name,
        items,
        total_amount
    )

    orders.append(new_order)

    save_orders()


    print("\nOrder placed successfully!")
    print("Order ID:", order_id)
    print("Status:", new_order.status)

def view_all_orders():

    print("\n============================================================")
    print("                    ALL ORDERS")
    print("============================================================")


    if len(orders) == 0:

        print("No orders available.")

        return
        
    print(
        f"{'ID':<6}"
        f"{'Customer':<20}"
        f"{'Items':<8}"
        f"{'Amount':<15}"
        f"{'Status':<15}"
    )

    print("-" * 64)


    for order in orders:

        if order.status.lower() == "delivered":

            print(
                f"*** {order.order_id:<3}"
                f"{order.customer_name:<20}"
                f"{len(order.items):<8}"
                f"Rs {order.total_amount:<11.2f}"
                f"{order.status:<15} ***"
            )

        else:

            print(
                f"{order.order_id:<6}"
                f"{order.customer_name:<20}"
                f"{len(order.items):<8}"
                f"Rs {order.total_amount:<11.2f}"
                f"{order.status:<15}"
            )


    print("-" * 64)

    print("*** = Delivered Order ***")

def search_order():

    print("\n================================")
    print("        SEARCH ORDER")
    print("================================")


    try:

        order_id = int(
            input("Enter Order ID: ")
        )

    except ValueError:

        print("Error: Order ID must be a number.")

        return


    for order in orders:

        if order.order_id == order_id:

            print("\nOrder Found!")
            print("----------------------------")
            print("Order ID:", order.order_id)
            print("Customer:", order.customer_name)
            print("Items:", ", ".join(order.items))
            print("Item Count:", len(order.items))
            print(
                "Total Amount: Rs",
                f"{order.total_amount:.2f}"
            )
            print("Status:", order.status)
            print("----------------------------")

            return

    print("No order found with ID:", order_id)

load_orders()

while True:

    print("\n")
    print("========================================")
    print("   FOOD DELIVERY ORDER MANAGEMENT")
    print("========================================")
    print("1. Place New Order")
    print("2. View All Orders")
    print("3. Search Order by ID")
    print("4. Exit")
    print("========================================")


    choice = input(
        "Enter your choice: "
    ).strip()


    # Place Order

    if choice == "1":

        place_order()


    # View Orders

    elif choice == "2":

        view_all_orders()


    # Search Order

    elif choice == "3":

        search_order()


    # Exit

    elif choice == "4":

        print("\nThank you for using the Food Delivery System!")

        break


    # Invalid Choice

    else:

        print(
            "Invalid choice. Please enter 1, 2, 3 or 4."
        )
```
