```python
# 2. Write a function update_playlist_price(playlist, new_price) that updates the price of a given playlist in the playlist_prices dictionary. 
#    Test it by updating the price of any one playlist and printing the updated dictionary.

# Dictionary
playlist_prices = {
    "Chill Vibes": 99,
    "Workout Mix": 149,
    "Party Hits": 199,
    "Romantic Songs": 129,
    "Lofi Beats": 89
}
# Function to update price
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

# Update the price
update_playlist_price("Workout Mix", 179)

print("Updated Dictionary:")
print(playlist_prices)
```

```python
# 4. Given two sets: set1 contains the names of restaurants you have ordered from on Zomato, and set2 contains the names of restaurants
#   you have ordered from on Swiggy, find and print the union and intersection of these sets.<br><br><em><strong>Hint:</strong> Use the union() and 
#   intersection() methods of Python sets.</em>

# from Zomato
set1 = {"McDonald's", "Domino's", "KFC", "Pizza Hut"}

# from Swiggy
set2 = {"KFC", "Burger King", "Pizza Hut", "Subway"}

# Union
print("Union:")
print(set1.union(set2))

# Intersection
print("\nIntersection:")
print(set1.intersection(set2))
```
