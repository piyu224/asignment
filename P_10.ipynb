```python
# 1.

import pandas as pd

users = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "username": ["Rahul", "Priya", "Amit", "Neha", "Riya"],
    "city": ["Mumbai", "Delhi", "Pune", "Surat", "Ahmedabad"]
})

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7],
    "user_id": [101, 102, 101, 103, 105, 102, 101],
    "order_amount": [500, 750, 300, 900, 450, 600, 250]
})

result = pd.merge(users, orders, on="user_id", how="inner")

print(result)
```

```python
# 2.

import pandas as pd

users = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "username": ["Rahul", "Priya", "Amit", "Neha", "Riya"],
    "city": ["Mumbai", "Delhi", "Pune", "Surat", "Ahmedabad"]
})

orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7],
    "user_id": [101, 102, 101, 103, 105, 102, 101],
    "order_amount": [500, 750, 300, 900, 450, 600, 250]
})
result = pd.merge(users, orders, on="user_id", how="left")

print(result)
```

```python
# 3.

import pandas as pd

restaurants = pd.DataFrame({
    "restaurant_id": [101, 102, 103, 104, 105],
    "name": ["Dominos", "KFC", "Pizza Hut", "Burger King", "McDonalds"],
    "city": ["Mumbai", "Delhi", "Pune", "Surat", "Ahmedabad"]
})
zomato_reviews = pd.DataFrame({
    "review_id": [1, 2, 3, 4, 5],
    "restaurant_id": [101, 102, 101, 106, 103],
    "rating": [5, 4, 5, 3, 4],
    "reviewer_city": ["Mumbai", "Delhi", "Pune", "Nashik", "Pune"]
})

result = pd.merge(
    restaurants,
    zomato_reviews,
    on="restaurant_id",
    how="outer"
)

print(result)
```

```python
# 4.

import pandas as pd

playlists = pd.DataFrame({
    "playlist_id": [1, 2, 3, 4, 5],
    "user_id": [101, 102, 103, 104, 105],
    "genre": ["Pop", "Rock", "Bollywood", "Hip-Hop", "Classical"],
    "city": ["Mumbai", "Delhi", "Pune", "Surat", "Ahmedabad"]
})

user_profiles = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 106],
    "username": ["Rahul", "Priya", "Amit", "Neha", "Riya"],
    "city": ["Mumbai", "Delhi", "Mumbai", "Surat", "Ahmedabad"]
})

result = pd.merge(
    playlists,
    user_profiles,
    on=["user_id", "city"],
    how="inner"
)
print(result)
```

```python
# 5.

import pandas as pd

products = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "name": ["Laptop", "Mobile", "Headphones", "Keyboard"],
    "category": ["Electronics", "Electronics", "Audio", "Accessories"]
})

ratings = pd.DataFrame({
    "rating_id": [1, 2, 3, 4, 5],
    "product_id": [101, 102, 103, 105, 102],
    "rating": [5, 4, 5, 3, 4]
})

result = pd.merge(
    products,
    ratings,
    on="product_id",
    how="right"
)

print(result)
```
