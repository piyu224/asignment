```python
# 2

import pandas as pd

df = pd.DataFrame({
    "product_id": ["P101", "P102", "P101", "P103", "P102", "P104"],
    "user_id": ["U01", "U02", "U01", "U03", "U02", "U04"],
    "rating": [5, 4, 5, 3, 4, 5],
    "review_text": ["Good product","Nice quality","Good product","Average","Nice quality","Excellent"]
})

print("Original DataFrame:")
print(df)

duplicate_count_before = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("\nDuplicate reviews before drop_duplicates():", duplicate_count_before)

df_clean = df.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)

duplicate_count_after = df_clean.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Duplicate reviews after drop_duplicates():", duplicate_count_after)

print("\nDataFrame after removing duplicates:")
print(df_clean)
```

```python
import pandas as pd

df = pd.read_csv("Documents/Zomato-data-.csv")

print(df.columns)
```

```python
# 3. 

Q1 = df["approx_cost(for two people)"].quantile(0.25)
Q3 = df["approx_cost(for two people)"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["approx_cost(for two people)"] < lower) |
    (df["approx_cost(for two people)"] > upper)
]

print("Outlier Restaurants:")
print(outliers["name"])
```

```python
# 4

df = pd.DataFrame({
    "date": ["01-04-2026", "02-04-2026", "03-04-2026", "04-04-2026", "05-04-2026"],
    "tickets_sold": [100, 120, 110, 105, 1000]
})

from scipy.stats import zscore

df["z_score"] = zscore(df["tickets_sold"])

outliers = df[abs(df["z_score"]) > 2]

print("Outlier Dates:")
print(outliers["date"])
```

```python
# 5.

import pandas as pd
from scipy.stats import zscore

def find_outliers(df):
    z = zscore(df["transaction_amount"])
    return df[abs(z) > 2]

df = pd.DataFrame({
    "transaction_id": ["T101", "T102", "T103", "T104", "T105", "T106"],
    "transaction_amount": [100, 150, 120, 130, 110, 2000]
})

outliers = find_outliers(df)

print("Outlier Transactions:")
print(outliers)
```
