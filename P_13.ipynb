```python
# 2.

import pandas as pd

df = pd.read_csv("Documents/IPL_2023_Batsmen.csv")

mean_runs = df["Runs"].mean()
median_runs = df["Runs"].median()
mode_runs = df["Runs"].mode()[0]
minimum_runs = df["Runs"].min()
maximum_runs = df["Runs"].max()

print("Univariate Analysis of Runs")
print("----------------")
print("Mean Runs   :", mean_runs)
print("Median Runs :", median_runs)
print("Mode Runs   :", mode_runs)
print("Minimum Runs:", minimum_runs)
print("Maximum Runs:", maximum_runs)
```

```python
# 3.

import pandas as pd

data = {
    "name": ["Restaurant A", "Restaurant B", "Restaurant C",
             "Restaurant D", "Restaurant E"],
    "rating": [4.5, 4.2, 4.7, 4.0, 4.4],
    "votes": [1200, 850, 1500, 600, 1000],
    "avg_cost": [500, 400, 600, 350, 550]
}
df = pd.DataFrame(data)

print("Top 5 Zomato Restaurants:")
display(df)

correlation = df["rating"].corr(df["votes"])

print("Correlation between Rating and Votes:", correlation)
```
