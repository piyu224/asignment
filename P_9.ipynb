```python
# 1.

import pandas as pd

dates = pd.Series([
    "15-03-2025",
    "22-07-2024",
    "05-01-2026",
    "18-11-2023",
    "30-09-2025"
])

dates = pd.to_datetime(dates, format="%d-%m-%Y")
print(dates)
```

```python
# 2.

import pandas as pd

df = pd.DataFrame({
    "movie": ["Dangal", "Pathaan", "Jawan", "RRR"],
    "release_date": ["23-12-2016", "25-01-2023", "07-09-2023", "25-03-2022"]
})

df["release_date"] = pd.to_datetime(
    df["release_date"],
    format="%d-%m-%Y"
)

df["year"] = df["release_date"].dt.year
df["month"] = df["release_date"].dt.month
df["day"] = df["release_date"].dt.day

print(df)
```

```python
# 3.

import pandas as pd

dates = pd.date_range(
    start=pd.Timestamp.today(),
    periods=7,
    freq="D"
)
print(dates)
```

```python
# 4.

import pandas as pd

df = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "start_date": ["01-01-2026","15-02-2026","10-03-2026","25-04-2026","01-06-2026"]
})

df["start_date"] = pd.to_datetime(df["start_date"], format="%d-%m-%Y")

today = pd.Timestamp.today().normalize()

df["tenure_days"] = (today - df["start_date"]).dt.days

print(df)
```
